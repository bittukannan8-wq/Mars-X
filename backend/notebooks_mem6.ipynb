```python
import json

with open("mars_geology_simplified.geojson", encoding="utf-8") as f:
    geo = json.load(f)

units = set()

for feature in geo["features"]:
    unit = feature["properties"].get("unit")
    if unit is not None:
        units.add(str(unit))

print("Unique geological units:", len(units))
print("Unit names:", sorted(units))
```

```python
import json

with open("mars_geology_simplified.geojson", "r", encoding="utf-8") as f:
    geo = json.load(f)

features = geo["features"]

print("Total features:", len(features))

# Display all property names
properties = features[0]["properties"]

print("\nAvailable property names:")
for key in properties:
    print(key)

# Display the first five feature properties
print("\nFirst five feature properties:")

for feature in features[:5]:
    print(feature["properties"])
```

```python
# Check every property and its unique values

for key in features[0]["properties"]:
    values = set()

    for feature in features:
        value = feature["properties"].get(key)

        if value is not None:
            values.add(str(value).strip())

    print("\nProperty:", key)
    print("Unique values:", len(values))
    print("Sample values:", sorted(values)[:10])
```
